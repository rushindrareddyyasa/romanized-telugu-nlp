Download CMTET into shared Google Drive

```python
from pathlib import Path
import subprocess

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Romanized-Telugu-NLP"
)
CMTET_DIR = PROJECT_DIR / "data/raw/cmtet"

# Make sure the parent directory exists
CMTET_DIR.parent.mkdir(parents=True, exist_ok=True)

# Clone only if the destination does not already contain a repository
if (CMTET_DIR / ".git").exists():
    print("CMTET repository already exists.")
else:
    result = subprocess.run(
        [
            "git", "clone",
            "--depth", "1",
            "https://github.com/ksubbu199/cmtet-sentiment.git",
            str(CMTET_DIR)
        ],
        capture_output=True,
        text=True
    )

    if result.returncode == 0:
        print("CMTET downloaded successfully!")
    else:
        print("Download failed:")
        print(result.stderr)

# List downloaded files
if CMTET_DIR.exists():
    for path in CMTET_DIR.rglob("*"):
        if path.is_file():
            print(path.relative_to(PROJECT_DIR))
```

Download Dakshina

```python
from pathlib import Path
import subprocess

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Romanized-Telugu-NLP"
)
DAKSHINA_DIR = PROJECT_DIR / "data/raw/dakshina"

DAKSHINA_DIR.mkdir(parents=True, exist_ok=True)

ARCHIVE = DAKSHINA_DIR / "dakshina_dataset_v1.0.tar"

if ARCHIVE.exists() and ARCHIVE.stat().st_size > 0:
    print("Dakshina archive already exists.")
    print(f"Size: {ARCHIVE.stat().st_size / (1024**2):.2f} MB")
else:
    result = subprocess.run([
        "wget", "-c",
        "-O", str(ARCHIVE),
        "https://storage.googleapis.com/gresearch/dakshina/dakshina_dataset_v1.0.tar"
    ])

    if result.returncode == 0 and ARCHIVE.exists():
        print("Dakshina download completed.")
        print(f"Size: {ARCHIVE.stat().st_size / (1024**2):.2f} MB")
    else:
        print("Download failed or did not complete.")
```

Verify that the files are valid

```python
from pathlib import Path
import tarfile

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Romanized-Telugu-NLP"
)

# Paths to the downloaded datasets
cmtet_file = (
    PROJECT_DIR / "data/raw/cmtet/codemix_sentiment_data.txt"
)

dakshina_archive = (
    PROJECT_DIR / "data/raw/dakshina/dakshina_dataset_v1.0.tar"
)

print("========== CMTET CHECK ==========")

if cmtet_file.exists() and cmtet_file.stat().st_size > 0:
    print("File exists: YES")
    print(f"Size: {cmtet_file.stat().st_size:,} bytes")

    # Read a few lines without changing the file
    with cmtet_file.open(
        "r", encoding="utf-8", errors="replace"
    ) as file:
        for i, line in enumerate(file):
            if i == 5:
                break
            print(f"Line {i + 1}: {line.rstrip()}")
else:
    print("CMTET file missing or empty.")

print("\n========== DAKSHINA CHECK ==========")

if dakshina_archive.exists() and dakshina_archive.stat().st_size > 0:
    print(f"Archive size: {dakshina_archive.stat().st_size:,} bytes")

    try:
        with tarfile.open(dakshina_archive, "r:") as tar:
            members = tar.getmembers()
            print("Valid TAR archive: YES")
            print("Total archive entries:", len(members))
            print("\nFirst 10 archive paths:")

            for member in members[:10]:
                print(member.name)

    except (tarfile.TarError, OSError) as error:
        print("Archive validation failed:", error)
else:
    print("Dakshina archive missing or empty.")
```
